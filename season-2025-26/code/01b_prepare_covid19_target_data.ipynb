```python
import numpy as np
import pandas as pd
import os
from pathlib import Path
import matplotlib.pyplot as plt
```

```python
path_folder = "../target-data/CovidHub/DelphiAPI/"

# List only directories inside path_folder
files = os.listdir(path_folder)
print(files)
```

```python
df_locations = pd.read_csv("https://raw.githubusercontent.com/cdcepi/FluSight-forecast-hub/main/auxiliary-data/locations.csv")
df_locations
```

```python
import shutil

for file in files:
    df = pd.read_csv(path_folder + file, index_col = 0)
    df['geo_value'] = df['geo_value'].str.upper()
    df.rename(columns = {'geo_value': 'abbreviation', 'target_end_date': 'week_ending_date'}, inplace=True)
    df = pd.merge(df, df_locations[['abbreviation', 'location', 'location_name']], on = 'abbreviation')
    df = df[['week_ending_date', 'location', 'location_name', 'value']]
    dest = f"../target-data/CovidHub/"
    new_filename = file.split('covid_hosp26_')[1] 
    new_filename2 = new_filename.split('.csv')[0]
    new_filename3 = new_filename2 + '_covid_target_hospital_admissions.csv'
    df.to_csv(dest + new_filename3)
```

```python
path_folder_saved = "../target-data/CovidHub/DelphiAPI-formatted/"
# List only directories inside path_folder
new_files = os.listdir(path_folder_saved)

df = pd.read_csv(path_folder_saved + new_files[2], index_col = 0)
df['week_ending_date'] = pd.to_datetime(df['week_ending_date'])
df = df.sort_values(by = ['location', 'week_ending_date'])
df_US = df[df['location'] == 'US']
plt.figure(figsize=(10,4), dpi=200)
plt.scatter(df_US['week_ending_date'], df_US['value'], s= 5)
plt.show()
```
