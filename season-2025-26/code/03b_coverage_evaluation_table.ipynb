```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import warnings
warnings.filterwarnings("ignore")
```

```python
challenge = "CovidHub"
df_scores = pd.read_csv(f"../model-evaluation/forecast_scores/forecast_scores_wis_ae_coverage_log_{challenge}.csv")
df_scores.head()
```

```python
df_scores[(df_scores['reference_date']> '2026-05-23')].model.unique()
```

```python
len(df_scores.reference_date.unique())
```

```python
df_scores = df_scores[(df_scores['reference_date']>= '2025-11-22') & (df_scores['reference_date']<= '2026-05-23')]
```

```python
df_scores
```

```python
df_scores['interval_coverage_50'].replace({True:1, False:0}, inplace=True)
df_scores['interval_coverage_90'].replace({True:1, False:0}, inplace=True)
```

```python
df_cov = df_scores.groupby(['model', 'location', 'reference_date', 'horizon'])[['interval_coverage_50', 'interval_coverage_90']].mean()
df_cov
```

```python
df_cov_final = df_cov.groupby(['model'])[['interval_coverage_50', 'interval_coverage_90']].mean()
df_cov_final
```

```python
df_cov_final.head(50)
```

```python
df_cov_final.to_csv(f"../model-evaluation/evaluation-table/coverage_table_log_{challenge}.csv")
```

```python
df_cov_final['dist_50'] = abs(df_cov_final['interval_coverage_50'] - 0.50)
df_cov_final = df_cov_final.sort_values('dist_50')
df_cov_final['position'] = [i+1 for i in range(len(df_cov_final))]
df_cov_final.head(60)
```

```python
df_cov_final['dist_90'] = abs(df_cov_final['interval_coverage_90'] - 0.90)
df_cov_final = df_cov_final.sort_values('dist_90')
df_cov_final['position'] = [i+1 for i in range(len(df_cov_final))]
df_cov_final.head(60)
```
