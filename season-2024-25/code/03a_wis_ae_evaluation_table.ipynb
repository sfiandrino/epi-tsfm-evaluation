```python
import pandas as pd
```

```python
challenge = 'CovidHub'
df_eval_loc = pd.read_csv(f"../model-evaluation/forecast_pairwise_comparison_scores/forecast_scores_pairwise_comparison_summary_2024-2025_log_{challenge}.csv")
df_eval_loc.head()
```

```python
df_eval_loc.model.unique()
```

```python
df_eval_loc_cleaned = (
    df_eval_loc[["model", "location", "wis_scaled_relative_skill", "ae_median_scaled_relative_skill"]]
    .drop_duplicates(subset=["model", "location"])
    .reset_index(drop=True)
)
df_eval_loc_cleaned
```

```python
df_eval_median = (
    df_eval_loc_cleaned
    .groupby("model", as_index=False)
    [["wis_scaled_relative_skill", "ae_median_scaled_relative_skill"]]
    .median()
)
df_eval_median
```

```python
# order df_eval_cleaned by wis_scaled_relative_skill from lowest to highest
df_eval_median_sorted = df_eval_median.sort_values(by="wis_scaled_relative_skill")
df_eval_median_sorted.to_csv(f"../model-evaluation/evaluation-table/df_evaluation_{challenge}_2024-2025_log.csv", index=False)
```

```python
df_eval_median_sorted['position'] = [i+1 for i in range(len(df_eval_median_sorted))]
df_eval_median_sorted.tail(60)
```

```python

# order df_eval_cleaned by wis_scaled_relative_skill from lowest to highest
df_eval_median_sorted = df_eval_median.sort_values(by="ae_median_scaled_relative_skill")
df_eval_median_sorted['position'] = [i+1 for i in range(len(df_eval_median_sorted))]
df_eval_median_sorted.head(60)
```
