```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import warnings
warnings.filterwarnings("ignore")
```

```python
challenge = "CovidHub"
df_scores = pd.read_csv(f"../model-evaluation/forecast_scores/forecast_scores_wis_ae_coverage_2024-2025_log_{challenge}.csv")
df_scores.head()
```

```python
df_scores['interval_coverage_50'].replace({True:1, False:0}, inplace=True)
df_scores['interval_coverage_90'].replace({True:1, False:0}, inplace=True)
```

```python
df_scores = df_scores[(df_scores['reference_date']>= '2024-11-30') & (df_scores['reference_date']<= '2025-05-24')]  # evaluation window
df_cov = df_scores.groupby(['model', 'location', 'reference_date', 'horizon'])[['interval_coverage_50', 'interval_coverage_90']].mean()
df_cov_final = df_cov.groupby(['model'])[['interval_coverage_50', 'interval_coverage_90']].mean()
```

```python
df_cov_final.head(50)
```

```python
df_cov_final.to_csv(f"../model-evaluation/evaluation-table/coverage_{challenge}_2024-2025_log.csv")
```

```python
df_cov_final['dist_50'] = abs(df_cov_final['interval_coverage_50'] - 0.50)
df_cov_final = df_cov_final.sort_values('dist_50')
df_cov_final['position'] = [i+1 for i in range(len(df_cov_final))]
df_cov_final.tail(60)
```

```python
df_cov_final['dist_90'] = abs(df_cov_final['interval_coverage_90'] - 0.90)
df_cov_final = df_cov_final.sort_values('dist_90')
df_cov_final['position'] = [i+1 for i in range(len(df_cov_final))]
df_cov_final.head(60)
```
